```python
speeches_df = df.drop(columns={"notes", "url", "error"}).dropna()
speeches_df.head(5)
```

```python
speeches_df.info()
```

```python
speeches_df['date'] = pd.to_datetime(speeches_df['date'], format='mixed', errors='coerce')
speeches_df = speeches_df.dropna(subset=['date'])
speeches_df.info()
```

```python
speeches_df = speeches_df.rename(columns={"body": "text"}).set_index("date").sort_index()
speeches_df.head(5)
```
